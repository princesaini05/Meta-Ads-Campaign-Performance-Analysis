```python
#convert dates
```

```python
df['start_date'] = pd.to_datetime(df['start_date'])
df['end_date']   = pd.to_datetime(df['end_date'])
```

```python
# Create a new column: how many days did the campaign run?
```

```python
df['duration_days'] = (df['end_date'] - df['start_date']).dt.days
```

```python
# Summary statistics
```

```python
df[['spend','revenue','clicks','leads','purchases','CTR','CPC','CVR','ROI','ROAS']].describe().round(2)
```

```python
# fix "Nan" strings
```

```python
df['region'] = df['region'].replace('Nan', np.nan)
df['objective'] = df['objective'].replace('Nan', np.nan)
df['placement'] = df['placement'].replace('Nan', np.nan)
df['adset'] = df['adset'].replace('Nan', np.nan)
```

```python
# confirm no duplicate campaign IDs
```

```python
print("Duplicate IDs:", df['campaign_id'].duplicated().sum())
```

```python
#overall business KPIs
```

```python
total_spend   = df['spend'].sum()
total_revenue = df['revenue'].sum()
total_leads   = df['leads'].sum()
total_purch   = df['purchases'].sum()
overall_roi  = ((total_revenue - total_spend) / total_spend) * 100
overall_roas = total_revenue / total_spend

print(f"Total Spend:    ₹{total_spend:,.2f}")
print(f"Total Revenue:  ₹{total_revenue:,.2f}")
print(f"Total Leads:    {total_leads:,}")
print(f"Total Purchases:{total_purch:,}")
print(f"Overall ROI:    {overall_roi:.2f}%")
print(f"Overall ROAS:   {overall_roas:.2f}x")
```

```python
#how much money is stuck where?
```

```python
decision_summary= df.groupby('Decision').agg({'campaign_id':'count','spend':'sum','revenue':'sum','leads':'sum','purchases':'sum'}).rename(columns={'campaign_id':'campaigns'})

decision_summary['ROI_%'] = ((decision_summary['revenue'] - decision_summary['spend'])/ decision_summary['spend'] * 100).round(2)

print(decision_summary)
```

```python
#Top 10 Performers by ROI
```

```python
top_roi = df.nlargest(10,'ROI')[['campaign_id','campaign_name','region','platform','adset','spend','revenue','ROI','ROAS','CPC','CVR']]
top_roi = top_roi.sort_values(by='revenue', ascending=False)

print(top_roi)
```

```python
#Worst Performers
```

```python
#High spend + negative ROI
```

```python
red_flags = df[(df['spend']>4000)&(df['ROI']<0)].sort_values('spend',ascending=False)

print((red_flags[['campaign_id','campaign_name','region','spend','revenue','ROI','Decision']]).head(15))
print(f"\n Wsate spend on red flags: ₹{red_flags['spend'].sum():,.2f}")
```

```python
#zero-purchase campaigns
```

```python
zero_pur = df[(df['purchases'] == 0) & (df['spend'] > 2000)]

print(zero_pur[['campaign_id','campaign_name','region','spend','revenue','purchases','ROI','Decision']])
print(f"Campaigns with 0 purchases & spend > ₹2000: {len(zero_pur)}")
print(f"Total wasted: ₹{zero_pur['spend'].sum():,.2f}")
```

```python
#Multi-Level Analysis
```

```python
#Platform
```

```python
platform_perf = df.groupby('platform').agg({'spend':'sum','revenue':'sum','clicks':'sum','purchases':'sum','CTR':'mean','CPC':'mean','ROI':'mean'}).round(2)

print(platform_perf)
```

```python
#Region
```

```python
region_perf = df.groupby('region').agg({'spend':'sum','revenue':'sum','ROI':'mean','CPC':'mean','CVR':'mean'}).round(2).sort_values('ROI', ascending=False)
print(region_perf)
```

```python
#Ad Set
```

```python
adset_perf = df.groupby('adset').agg({'spend':'sum','revenue':'sum','CTR':'mean','CVR':'mean','ROI':'mean'}).round(2).sort_values('ROI', ascending=False)

print(adset_perf)
```

```python
#Creative Type
```

```python
creative_perf = df.groupby('creative_type').agg({'spend':'sum','revenue':'sum','CTR':'mean','ROI':'mean'}).round(2).sort_values('ROI', ascending=False)

print(creative_perf)
```

```python
#Age × Gender
```

```python
age_gender = df.pivot_table(values='ROI', index='age_group',columns='gender', aggfunc='mean').round(2)

print(age_gender)
```

```python
#Device & Placement
```

```python
print("Device:\n", df.groupby('device')[['CTR','CVR','ROI']].mean().round(2))
print("\nPlacement:\n", df.groupby('placement')[['CTR','CVR','ROI']].mean().round(2))
```

```python
with pd.ExcelWriter("D:\DATA ANALYSIS\Meta Ads Campaign Performance Analysis\Outputs\excel files\Meta_Ads_Report.xlsx", engine='openpyxl') as writer:
    df.to_excel(writer, sheet_name='Cleaned_Data', index=False)
    decision_summary.to_excel(writer, sheet_name='Decision_Summary')
    platform_perf.to_excel(writer, sheet_name='Platform_Perf')
    region_perf.to_excel(writer, sheet_name='Region_Perf')
    adset_perf.to_excel(writer, sheet_name='Adset_Perf')
    creative_perf.to_excel(writer, sheet_name='Creative_Perf')
    top_roi.to_excel(writer, sheet_name='Top_10_ROI', index=False)
    red_flags.to_excel(writer, sheet_name='Red_Flags', index=False)
```
