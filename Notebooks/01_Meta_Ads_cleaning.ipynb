```python
import pandas as pd
```

```python
import numpy as np
```

```python
# Load data
df = pd.read_csv(r"D:\DATA ANALYSIS\Meta Ads Campaign Performance Analysis\Data\Raw\Meta_Ads_Dataset.csv")
```

```python
df
```

```python
print(df.shape)
```

```python
print(df.info())
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
text_cols = ['campaign_name','adset','platform','objective','status',
             'country','region','age_group','gender','placement',
             'device','creative_type']

for col in text_cols:
    df[col] = df[col].astype(str).str.strip().str.title()
```

```python
df['status'] = df['status'].str.upper()
```

```python
df
```

```python
#Fix Date Columns
```

```python
def parse_date(x):
    for fmt in ('%Y-%m-%d','%Y/%m/%d','%m-%d-%Y','%d/%m/%Y'):
        try:
            return pd.to_datetime(x, format=fmt)
        except:
            continue
    return pd.NaT

df['start_date'] = df['start_date'].apply(parse_date)
df['end_date']   = df['end_date'].apply(parse_date)
```

```python
#Handle Missing Values
```

```python
df['region'] = df['region'].replace('Nan', np.nan)
df['objective'] = df['objective'].replace('', np.nan)
```

```python
print(df.isnull().sum())
```

```python
df[df['region'].isnull()]
```

```python
#Remove Duplicates
```

```python
df = df.drop_duplicates(subset='campaign_id', keep='first')
```

```python
#Fix Numeric Types
```

```python
#First look at the data
```

```python
num_cols = ['impressions','reach','clicks','spend','leads',
            'purchases','revenue','conversion_value']
df[num_cols] = df[num_cols].apply(pd.to_numeric, errors='coerce')
```

```python
#Create All KPIs
```

```python
#CTR
df['CTR']  = (df['clicks'] / df['impressions']) * 100
```

```python
#CPC
df['CPC']  = df['spend'] / df['clicks']
```

```python
#CPA
df['CPA']  = df['spend'] / df['purchases'].replace(0, np.nan)
```

```python
#CVR
df['CVR']  = (df['purchases'] / df['clicks']) * 100
```

```python
#CPL
df['CPL']  = df['spend'] / df['leads'].replace(0, np.nan)
```

```python
#ROI
df['ROI']  = ((df['revenue'] - df['spend']) / df['spend']) * 100
```

```python
#ROAS
df['ROAS'] = df['revenue'] / df['spend']
```

```python
#Frequency
df['Frequency'] = df['impressions'] / df['reach']
```

```python
#Round for Readability
```

```python
df[['CTR','CPC','CPA','CVR','CPL','ROI','ROAS','Frequency']] = \
    df[['CTR','CPC','CPA','CVR','CPL','ROI','ROAS','Frequency']].round(2)
```

```python
df
```

```python
##Handle Missing Values
```

```python
# Drop rows where key metrics are missing AND can't be imputed
```

```python
df=df.dropna(subset=['impressions','clicks','spend'])
```

```python
# Fill missing leads/purchases/revenue with 0
```

```python
df['leads']    = df['leads'].fillna(0)
df['purchases']= df['purchases'].fillna(0)
df['revenue']  = df['revenue'].fillna(0)
```

```python
# Fill missing reach with impressions
df['reach'] = df['reach'].fillna(df['impressions'])
```

```python
print(df[['reach', 'leads', 'purchases', 'revenue']].isnull().sum())
```

```python
#Remove Invalid & Sentinel Values
```

```python
df=df[(df['spend']>=0)&
    (df['revenue']>=0)&
    (df['leads']>=0)&
    (df['purchases']>=0)]
```

```python
#Remove sentinel reach
```

```python
df = df[df['reach'] < 10_000_000]
```

```python
#Classification Logic (Continue / Pause / Discontinue)
```

```python
def decision(row):
    if pd.isna(row['ROI']) or pd.isna(row['CVR']):
        return 'Review'
    if row['ROI'] > 1 and row['CVR'] > 1:
        return 'Continue'
    elif row['ROI'] > 0:
        return 'Pause'
    else:
        return 'Discontinue'

df['Decision'] = df.apply(decision, axis=1)

print(df['Decision'].value_counts())
```

```python
#Save Cleaned Data
```

```python
df.to_csv("D:/DATA ANALYSIS/Meta Ads Campaign Performance Analysis/Data/Cleaned files/meta_ads_cleaned.csv", index=False)
df.to_excel("D:/DATA ANALYSIS/Meta Ads Campaign Performance Analysis/Outputs/meta_ads_final_for_powerbi.xlsx", index=False)

print("Files saved ✅")
```
